Print all departure and arrival airports (using format *CITY (CODE)*) where arrival airport is *Grozny* and scheduled departure was on 19.07.2017.

HINT: you can use `to_char(f.scheduled_departure, 'DD.MM.YYYY')` to work with dates.

The output should look as follows:
```
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')

('19.07.2017', 'Moscow (DME)', 'Grozny (GRV)')
```

````python
cursor = conn.cursor()

query = ("""
    SELECT   to_char(f.scheduled_departure, 'DD.MM.YYYY') AS when,
             d.city || ' (' || f.departure_airport || ')' AS departure,
             a.city || ' (' || f.arrival_airport || ')' AS arrival
    FROM     ticket_flights tf
             JOIN flights f ON tf.flight_id = f.flight_id
			 JOIN airports a ON f.arrival_airport = a.airport_code
			 JOIN airports d ON f.departure_airport = d.airport_code
	WHERE a.city = 'Grozny' AND
			to_char(f.scheduled_departure, 'DD.MM.YYYY') = '19.07.2017'
""")

cursor.execute(query)

for line in cursor:
  print(f"{line}")

cursor.close()
````

For each aircraft type (aircraft_code, model) and seat type (fare_conditions) print its code and model, number of different seats and their names (for the latter use function string_agg). Order the output by

````python
cursor = conn.cursor()

query = ("""
    SELECT   a.aircraft_code,
             a.model,
             s.fare_conditions,
             COUNT(s.seat_no),
             string_agg(s.seat_no, ', ')
    FROM     aircrafts a
             JOIN seats s ON a.aircraft_code = s.aircraft_code
    GROUP BY a.aircraft_code, a.model, s.fare_conditions
    ORDER BY COUNT(s.seat_no) ASC;
""")

cursor.execute(query)

for line in cursor:
  print(f"{line}")

cursor.close()
````
